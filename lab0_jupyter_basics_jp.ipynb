## 1. セル(Cell)とは

Jupyter Notebookは「セル」と呼ばれる小さなブロックの集まりでできています。セルには
主に2種類あります:

- **コードセル** — Pythonのコードを書いて実行する場所(このすぐ下のセルがそうです)
- **Markdownセル** — 説明文を書く場所(今あなたが読んでいる、この文章自体がMarkdown
  セルです)

**セルの実行方法:** セルを選んだ状態で **Shift + Enter** を押してください。
(またはセル左上の▶ボタンをクリックしてもかまいません。)

下のセルを選んで、実際に実行してみてください。

```
print("こんにちは、Jupyter Notebook!")
```

実行すると、セルの左側に `[1]` のような数字が表示されたはずです。これは
「このセルが何番目に実行されたか」を示す番号です。次の章で、この番号が実は
とても大事な意味を持つことを説明します。

## 3. カーネル(Kernel)とは

**カーネル**とは、あなたが実行したコードを裏side で実際に動かしている、Pythonの
プログラムそのものです。

> 💡 **たとえ話:** Notebookの画面(セルが並んでいる部分)が「注文書」だとしたら、
> カーネルは注文を実際に調理する「厨房」です。注文書に書いた内容(コード)は、
> 厨房(カーネル)に送られて初めて実行されます。

重要なのは、**カーネルは「今までに実行したセルの内容」をすべて覚えている**という
ことです。上で定義した `price` や `total` は、今もカーネルの中に残っています。
試してみましょう。

## 4. 実行順序の重要な注意点

ここが、初めてNotebookを使う人が一番つまずきやすいポイントです。

**Notebookは「上から下へ書かれている」ように見えますが、カーネルは
「実際にあなたが実行した順番」しか知りません。** 見た目の順番と、実行した順番は、
**別物**です。

次の2つのセルで、これを実際に体験してみましょう。

```
# セルA: xに10を入れる
x = 10
print("x =", x)
```

```
# セルB: xを使って計算する
print("x × 2 =", x * 2)
```

上の「セルA」→「セルB」の順に実行すれば、`x × 2 = 20` と表示されます。

**ここで試してほしいこと:** セルAをもう一度開いて、`x = 10` を `x = 999` に書き
換えて実行し、その後で**セルBをもう一度実行せずに**、この文章を読んでいる状態を
想像してください。画面上ではセルBの出力はまだ `20` のままです。しかし、
セルBを再実行すると `1998` に変わります。

**つまり、画面に表示されている数字が、必ずしも「今の状態」を正しく表しているとは
限りません。** 上から下に読むだけでなく、「自分が本当は何を、どの順番で実行したか」
を意識する習慣が、この後のモジュール1〜3でとても重要になります。

## 5. カーネルの再起動(Restart Kernel)

「実行した順番が分からなくなってしまった」「変数の中身がおかしい気がする」——
そんなときは、カーケルを**再起動**して、まっさらな状態からやり直すことができます。

再起動すると、それまでにカーネルが覚えていた変数はすべて消えます。その代わり、
上のメニューから **「Restart Kernel and Run All Cells」** を選べば、Notebookの
セルを**上から順番に**すべて実行し直してくれます。これにより、「見た目の順番」と
「実行の順番」を一致させることができます。

> ⚠️ **この後のモジュールで実際に使います:** モジュール1・3のボーナスセクションで、
> 新しいライブラリを読み込んだ際に「カーネルの再起動」が必要になる場面が実際に
> 出てきます。そのときは、今説明したこの操作を思い出してください。

## まとめ

- **セル**はコード(またはMarkdown)を書く単位。**Shift + Enter**で実行する
- **カーネル**は、実行したコードを裏で動かしているPythonプロセスで、変数の中身を
  覚えている
- **実行した順番**が本当の状態を決める。**見た目の上下の順番ではない**
- 分からなくなったら、**「Restart Kernel and Run All Cells」**でやり直せる
